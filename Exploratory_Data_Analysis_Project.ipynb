# Exploratory Data Analysis (EDA) Project

**Internship Task:** Exploratory Data Analysis (EDA)

## Objective
Analyze a sales dataset to uncover patterns, trends, relationships, and important factors influencing sales and profit.


## 1. Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
pd.set_option("display.max_columns", None)
print("Libraries imported successfully.")
```

## 2. Load the Dataset

```python
df = pd.read_csv("sales_eda_dataset.csv", parse_dates=["Order_Date"])
print("Dataset shape:", df.shape)
display(df.head())
```

## 3. Initial Inspection

```python
df.info()

print("\nStatistical summary:")
display(df.describe(include="all").T)

print("\nMissing values:")
display(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())
```

## 4. Data Cleaning

```python
clean_df = df.copy()

for col in clean_df.select_dtypes(include="number").columns:
    clean_df[col] = clean_df[col].fillna(clean_df[col].median())

clean_df = clean_df.drop_duplicates().reset_index(drop=True)

print("Shape after cleaning:", clean_df.shape)
print("Remaining missing values:", clean_df.isnull().sum().sum())
```

## 5. Feature Preparation

```python
clean_df["Month"] = clean_df["Order_Date"].dt.to_period("M").astype(str)
clean_df["Profit_Margin"] = clean_df["Profit"] / clean_df["Sales"]
display(clean_df.head())
```

## 6. Sales by Region

```python
region_sales = clean_df.groupby("Region")["Sales"].sum().sort_values(ascending=False)
display(region_sales)

plt.figure(figsize=(8,5))
sns.barplot(x=region_sales.index, y=region_sales.values)
plt.title("Total Sales by Region")
plt.xlabel("Region")
plt.ylabel("Sales")
plt.tight_layout()
plt.show()
```

## 7. Sales by Category

```python
category_sales = clean_df.groupby("Category")["Sales"].sum().sort_values(ascending=False)
display(category_sales)

plt.figure(figsize=(9,5))
sns.barplot(x=category_sales.index, y=category_sales.values)
plt.title("Total Sales by Product Category")
plt.xlabel("Category")
plt.ylabel("Sales")
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

## 8. Monthly Sales Trend

```python
monthly_sales = clean_df.groupby("Month")["Sales"].sum()

plt.figure(figsize=(12,5))
sns.lineplot(x=monthly_sales.index, y=monthly_sales.values, marker="o")
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 9. Monthly Profit Trend

```python
monthly_profit = clean_df.groupby("Month")["Profit"].sum()

plt.figure(figsize=(12,5))
sns.lineplot(x=monthly_profit.index, y=monthly_profit.values, marker="o")
plt.title("Monthly Profit Trend")
plt.xlabel("Month")
plt.ylabel("Profit")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 10. Sales Distribution

```python
plt.figure(figsize=(9,5))
sns.histplot(clean_df["Sales"], bins=30, kde=True)
plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

## 11. Profit Outliers

```python
plt.figure(figsize=(9,5))
sns.boxplot(x=clean_df["Profit"])
plt.title("Profit Distribution and Potential Outliers")
plt.xlabel("Profit")
plt.tight_layout()
plt.show()
```

## 12. Discount vs Sales

```python
plt.figure(figsize=(9,5))
sns.scatterplot(data=clean_df, x="Discount", y="Sales", hue="Category", alpha=0.7)
plt.title("Discount vs Sales")
plt.xlabel("Discount")
plt.ylabel("Sales")
plt.tight_layout()
plt.show()
```

## 13. Customer Rating by Category

```python
plt.figure(figsize=(9,5))
sns.boxplot(data=clean_df, x="Category", y="Customer_Rating")
plt.title("Customer Rating by Category")
plt.xlabel("Category")
plt.ylabel("Customer Rating")
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

## 14. Correlation Analysis

```python
numeric = clean_df.select_dtypes(include="number")
corr = numeric.corr()

plt.figure(figsize=(10,7))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", center=0)
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()

display(corr["Sales"].sort_values(ascending=False).to_frame("Correlation with Sales"))
```

## 15. Key Findings

```python
print("Highest-sales region:", region_sales.index[0])
print("Highest-sales category:", category_sales.index[0])
print("Highest-sales month:", monthly_sales.idxmax())
print("Highest monthly sales:", round(monthly_sales.max(), 2))

print("\nThe analysis compares regional and category performance, identifies monthly trends, examines distributions and potential outliers, and measures relationships between numerical variables.")
```
